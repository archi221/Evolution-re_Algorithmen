```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import glob
import re

FITNESS_THRESHOLD = 0.1

re_pattern = ".\\/DE_(.*)_(.*)_results"
```

```python
DATA_DIR = '.'
re_pattern = ".\\/DE_(.*)_(.*)_results"
pattern = os.path.join(DATA_DIR, f'DE_*_*_results*')
files = glob.glob(pattern)
#print(files)
ackley = []
sphere = []
for file in files:
    data = pd.read_csv(file, sep=",")
    data_thresh = data[data[" fitness"] >= FITNESS_THRESHOLD]
    
    result = re.match(re_pattern, file)
    funktion, strategie = result.groups(1)
    if funktion == "sphere":
        sphere.append((strategie, data_thresh))
    elif funktion == "ackley":
        ackley.append((strategie, data_thresh))
    else:
        print(f"unkown funktion: {funtkion}")
```

```python
DATA_DIR = "../1/final_solution"
pattern = os.path.join(DATA_DIR, f'*_Funktion.csv')
files = glob.glob(pattern)
ackley = []
sphere = []
for file in files:
    data = pd.read_csv(file, sep=",")
    data_thresh = data[data[" fitness"] >= FITNESS_THRESHOLD]
    
    result = re.match(re_pattern, file)
    funktion, strategie = result.groups(1)
    if funktion == "sphere":
        sphere.append((strategie, data_thresh))
    elif funktion == "ackley":
        ackley.append((strategie, data_thresh))
    else:
        print(f"unkown funktion: {funtkion}")
print(files)
```

```python
ackley_string = ""
sphere_string = ""

for (a_strategie, a_data), (s_strategie, s_data) in zip(ackley, sphere):
    a_start = f"{a_data[' fitness'][0]:.2f}"
    s_start = f"{s_data[' fitness'][0]:.2f}"
    
    ackley_string += f"{a_strategie:18} start: {a_start:>8} | Generationen: {len(a_data):>4}\n"
    sphere_string += f"{s_strategie:18} start: {s_start:>8} | Generationen: {len(a_data):>4}\n"

print("=== ACKLEY ===")
print(ackley_string)
print("\n=== SPHERE ===")
print(sphere_string)
```

```python
fig, axes = plt.subplots()
for strategie, data in ackley:
    axes.plot(data[" fitness"], label=strategie)
    
plt.xlabel("Generations")
plt.ylabel("Fitness")
plt.title("Ackley funktion Optimization")
plt.legend()
plt.show()
```
